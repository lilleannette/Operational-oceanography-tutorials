# Searching the Copernicus Marine catalogue

Copernicus Marine offers hundreds of products, each containing several datasets. If you don't already know the exact dataset ID you need, the `copernicusmarine` Python package lets you browse the catalogue programmatically instead of searching the website.

This notebook walks through:
1. Fetching the full catalogue entry for a product
2. Listing the datasets within that product
3. Listing the variables available in each dataset

## 1. Fetch a product's catalogue entry

`copernicusmarine.describe()` downloads metadata for a given `product_id` and returns it as a structured object. Here we look up the **Global Ocean Biogeochemistry Analysis and Forecast** product (`GLOBAL_ANALYSISFORECAST_BGC_001_028`).

```python
# Run once per environment (e.g. once in Colab, or once in your local conda/venv)
%pip install -q copernicusmarine
```

```python
import copernicusmarine
from pprint import pprint

P_ID="GLOBAL_ANALYSISFORECAST_BGC_001_028"
catalogue = \
copernicusmarine.describe(product_id=P_ID, \
                          disable_progress_bar=True)

pprint(catalogue)
```

## 2. List the datasets in the product

A product typically bundles several datasets — e.g. different temporal resolutions (daily vs. monthly) or variable groups (biology, carbonate chemistry, nutrients, etc.). Loop through `catalogue.products[0].datasets` to see each dataset's ID.

```python
catalogue.products[0].title

for dataset in catalogue.products[0].datasets:
    print(dataset.dataset_id)
```

## 3. List the variables in each dataset

Each dataset contains one or more variables. `dataset.versions[0].parts[0].services[0].variables` gives the variable list for a dataset, including short name, standard name, and units — this is what you need to pick the right `dataset_id` and variable names for a `subset()` download request.

```python
dn=4 # dataset number - starting from 0
print(catalogue.products[0].datasets.index)
for dataset in catalogue.products[0].datasets:
    variables = dataset.versions[0].parts[0].services[0].variables
    print(variables,"\n")
    
```
